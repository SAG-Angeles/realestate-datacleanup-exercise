#### Ejercicio 09. ¿Son los precios promedios de "Valdemorillo" y "Galapagar" los mismos? (★★☆)

Imprime ambos promedios y escribe una conclusión sobre ellos.

```python
valdemorillo = ds[ds['level5'] == 'Valdemorillo']
galapagar = ds[ds['level5'] == 'Galapagar']


media_valdemorillo = valdemorillo['price'].mean()
media_galapagar = galapagar['price'].mean()


print(f"Precio promedio en Valdemorillo: {media_valdemorillo}")
print(f"Precio promedio en Galapagar: {media_galapagar}")


```

#### Ejercicio 16. Calcula la media y la varianza de muestra para las siguientes variables: precio, habitaciones, superficie y baños (debes usar el subconjunto obtenido del Ejercicio 14) (★★★)

Imprime ambos valores por cada variable.

```python
cinturon_sur = ['Fuenlabrada', 'Leganés', 'Getafe', 'Alcorcón']

ds_sur = ds[ds['level5'].isin(cinturon_sur)]

print("Precio")
print("Media:", ds_sur['price'].mean())
print("Varianza:", ds_sur['price'].var())

print("\nHabitaciones")
print("Media:", ds_sur['rooms'].mean())
print("Varianza:", ds_sur['rooms'].var())

print("\nSuperficie")
print("Media:", ds_sur['surface'].mean())
print("Varianza:", ds_sur['surface'].var())

print("\nBaños")
print("Media:", ds_sur['bathrooms'].mean())
print("Varianza:", ds_sur['bathrooms'].var())

```
